```python
# First Problrm (seperating strings from ints)

word = []
num = []
s = input("Enter a text containing both letters and numbers: ").strip()

for i in s:
    if i.isalpha():
        word.append(i)
    elif i.isdigit():
        num.append(int(i))
        
print("".join(word)+","+f"Sum:{sum(num)}")
```

```python
#P2:Checking the valid passwords

Check1 = 0
Check2 = 0

Password = input("Enter a password: ")

while len(Password)>=8 and len(Password)<=16:
    for i in Password:
        if not i.isalnum() and not i.isspace():
            Check1+=1
        if i.isupper():
            Check2+=1
        if Check1 >= 1 and Check2 >=1:
            print("Password is secure")
            Password = "0"
            break
    if Check1==0 and Check2==0:
        Password = input("Password must contain at least one uppercase letter and one special character! ")
```

```python
#P3
price_input = input("Enter the Price: ")
price = float(price_input)

member_input = input("Is the user a Member? (Yes/No): ")

if price < 50:
    discount_percent = 0
elif price <= 100:
    discount_percent = 10
else:
    discount_percent = 20

first_discount_amount = price * (discount_percent / 100)
intermediate_price = price - first_discount_amount

note = str(price) + " - " + str(discount_percent) + "% = " + str(intermediate_price)

if member_input == "Yes" or member_input == "yes":
    final_total = intermediate_price - (intermediate_price * 0.05)
   
    note = note + "; then " + str(intermediate_price) + " - 5% = " + str(final_total)
else:
    final_total = intermediate_price


print("Final Total: $" + str(final_total) + " (Note: " + note + ")")
```

```python
#P7
num = input("Enter a multi-digit number: ").strip()

explanation_steps = []

while len(num) != 1:
    
    math_text = "+".join(num)
    
    
    digit_sum = 0
    for digit in num:
        digit_sum += int(digit)
        
    explanation_steps.append(f"{math_text}={digit_sum}")
    

    num = str(digit_sum)


note_string = ", ".join(explanation_steps)

print(f"Digital Root: {num} ({note_string})")
```

```python
#P8
user_input = input("Enter numbers separated by commas (e.g., 1,1,2,3,3,3,4): ")

raw_list = user_input.split(",")

numbers = []
for item in raw_list:
    cleaned_item = item.strip()
    numbers.append(int(cleaned_item))

counts_dict = {}

for num in numbers:
    if num in counts_dict:
        counts_dict[num] = counts_dict[num] + 1
    else:
        counts_dict[num] = 1

output_list = []
for key in counts_dict:
    item_tuple = (key, counts_dict[key])
    output_list.append(item_tuple)

print("Output:", output_list)
```

```python
#P9
Sum = 0
F = 1

Num_input = input("Enter a number to calculate the factorial: ").strip()
Num = int(Num_input)

for i in range(1, Num + 1):
    F *= i
    Sum += i

if F % Sum == 0:
    print(f"Factorial is {F}, Sum is {Sum}. {F} is divisible by {Sum}.")
else:
    print(f"Factorial is {F}, Sum is {Sum}. {F} is not divisible by {Sum}.")
```

```python
#P10
Products = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
cart = []
total = 0

print("Welcome to Viyanna store ^^") 
while True:
    product = input("Add a product:  ").strip().lower()
    
    if product not in Products:
        print("Please enter an existing product!")
        continue
        
    cart.append(product)
    
    valid_choice = False
    while not valid_choice:
        add_more = input("Do you want to add more products? Enter yes or no: ").strip().lower()
        if add_more == 'yes':
            valid_choice = True
        elif add_more == 'no':
            valid_choice = True
        else:
            print("Please enter yes or no!")
            
    if add_more == 'no':
        break

for i in cart:
    total += Products[i]

print(f"The total price is ${total}")
```

```python
#P11
def Palindrome():
    W = input("Enter a word or a sentence: ").strip().lower()
    W_no_spaces = W.replace(" ", "")
    
    if W_no_spaces == W_no_spaces[::-1]:
        return True
    else:
        return False

print(Palindrome())
```

```python
#P13
user_input = input("Enter numbers separated by commas: ")
raw_list = user_input.split(",")

numbers = []
for item in raw_list:
    numbers.append(int(item.strip()))

peaks = []
for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

```python
#P14
user_input = input("Enter keys and values like A:1, B:2, C:1 >> ")
pairs = user_input.split(",")

data = {}
for pair in pairs:
    parts = pair.split(":")
    key = parts[0].strip()
    value = int(parts[1].strip())
    data[key] = value

flipped = {}
for key, value in data.items():
    if value not in flipped:
        flipped[value] = []
    flipped[value].append(key)

print(flipped)
```
